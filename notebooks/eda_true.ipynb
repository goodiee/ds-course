Exploratory data analysis with Pandas (Bike-sharing dataset)

This dataset contains daily bike rental records from a bike-sharing system. It includes information about date, season, weather conditions, and user counts, allowing analysis of how environmental and temporal factors affect bike usage.

Main Columns:

instant – record index

dteday – date of record

season – 1: spring, 2: summer, 3: fall, 4: winter

yr – year (0: 2011, 1: 2012)

mnth – month (1–12)

holiday – 1 if day is a holiday

weekday – day of the week

workingday – 1 if not weekend/holiday

weathersit – weather category (1 = good till 4 = severe)

temp – normalized temperature

atemp – normalized “feels like” temperature

hum – normalized humidity

windspeed – normalized wind speed

casual – number of casual users

registered – number of registered users

cnt – total rentals (target variable)

1. Count how many working days and non-working days exist in the dataset.

```python
data["workingday"].value_counts()
```

2. Average rentals on working vs non-working days

```python
avg_working = data[data["workingday"] == 1]["cnt"].mean()
avg_nonworking = data[data["workingday"] == 0]["cnt"].mean()

avg_working, avg_nonworking
```

3. Proportion of bad-weather days

```python
bad_weather_prop = (data["weathersit"] == 3).sum() / data.shape[0] # let's assume weathersit == 3 is bad
bad_weather_prop
```

```python
bad_weather_prop = data[data["weathersit"].isin([3, 4])].shape[0] / data.shape[0]
bad_weather_prop
```

4–5. Mean & std of rentals on high- vs low-demand days

```python
threshold = data["cnt"].median()

high = data[data["cnt"] > threshold]["cnt"]
low  = data[data["cnt"] <= threshold]["cnt"]

print("High demand mean:", round(high.mean(), 2))
print("High demand std:", round(high.std(), 2))

print("Low demand mean:", round(low.mean(), 2))
print("Low demand std:", round(low.std(), 2))
```

6. Are high-demand days only in specific seasons?

```python
high_seasons = data[data["cnt"] > threshold]["season"].unique()
high_seasons
```

all four seasons appear among the high-demand days

```python
season_map = {1: "winter", 2: "spring", 3: "summer", 4: "fall"}
[pd.Series(high_seasons).map(season_map)]
```

7. Rental stats for each season × workingday

```python
for (season, working), sub_df in data.groupby(["season", "workingday"]):
    print(f"Season: {season}, workingday: {working}")
    print(sub_df["cnt"].describe())
    print()
```

8. Which group has more high-demand days?

```python
high_working = (
    data[(data["workingday"] == 1) & (data["cnt"] > threshold)].shape[0] /
    data[data["workingday"] == 1].shape[0]
)

high_nonworking = (
    data[(data["workingday"] == 0) & (data["cnt"] > threshold)].shape[0] /
    data[data["workingday"] == 0].shape[0]
)

high_working, high_nonworking
```

9. Maximum number of rentals + count of such days + workingday share

```python
max_cnt = data["cnt"].max()
max_cnt
```

How many such days?

```python
n_days_max = data[data["cnt"] == max_cnt].shape[0]
n_days_max
```

What % are working days?

```python
working_share = (
    data[(data["cnt"] == max_cnt) & (data["workingday"] == 1)].shape[0]
    / n_days_max
)

working_share
```

10. Average rentals (cnt) for high vs low demand by season

```python
data["demand"] = np.where(data["cnt"] > threshold, "high", "low")
```

```python
data.groupby(["season", "demand"])["cnt"].mean()
```

```python
pd.crosstab(
    data["season"],
    data["demand"],
    values=data["cnt"],
    aggfunc="mean"
).T
```
